# Phase 6 — Experiments and Visualisations

Companion notebook for the article's experiments. This notebook is
an orchestration layer: it imports each experiment's `main()` and
regenerates every PNG plus the animated GIF. Running every cell from
top to bottom should produce the full set of figures used in the
article.

Map of experiments:

| ID | Topic                      | Output                                                |
|----|----------------------------|-------------------------------------------------------|
| A  | Prior → posterior          | `figures/exp_a_prior_to_posterior.png`                |
| B  | Four conjugate families    | `figures/exp_b_conjugate_families.png`                |
| C  | Sequential shrinkage       | `figures/exp_c_sequential_shrinkage.png`              |
| D  | Prior sensitivity          | `figures/exp_d_prior_sensitivity.png`                 |
| E  | Full FYF quarterly cycle   | `figures/exp_e_fyf_quarterly.png`                     |
| F  | Bayesian vs frequentist    | `figures/exp_f_bayesian_vs_frequentist.png`           |
| G  | Bayes factor vs AIC        | `figures/exp_g_model_comparison.png`                  |
| H  | Posterior predictive check | `figures/exp_h_posterior_predictive_check.png`        |
| —  | Animated posterior         | `figures/posterior_evolution.gif`                     |

Each script has a fixed seed and is a standalone module under
`scripts/`.

```python
from __future__ import annotations

import importlib
import sys
from pathlib import Path
from IPython.display import Image, display

REPO_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
FIG_DIR = REPO_ROOT / 'figures'
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
print(f'Repo root: {REPO_ROOT}')
print(f'Figure dir: {FIG_DIR}')
```

```python
# Helper: import a script by module name and call its main()
import importlib

def run_experiment(module_name: str) -> None:
    if module_name in sys.modules:
        del sys.modules[module_name]
    mod = importlib.import_module(module_name)
    mod.main()

def show_fig(name: str) -> None:
    path = FIG_DIR / name
    if not path.exists():
        print(f'(missing) {path}')
        return
    display(Image(filename=str(path)))
```

## A — prior to posterior

```python
run_experiment('scripts.exp_a_prior_to_posterior')
show_fig('exp_a_prior_to_posterior.png')
```

## B — four conjugate families

```python
run_experiment('scripts.exp_b_conjugate_families')
show_fig('exp_b_conjugate_families.png')
```
